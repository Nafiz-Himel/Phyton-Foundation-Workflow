Break and Continue Statements in Python

```python
sum = 0

for i in range(1,10):
    if i == 9:
        break
    elif i%2 == 0:
        continue
    sum+= i
    print(i,"is added to te sum..!")
print(sum)
```

C. Even, Odd, Positive and Negative

```python
i = int(input())
inpt = input();nums = inpt.split()

pos = 0; neg = 0; even = 0; odd = 0;

for j in range(0,i,1):
    x = int(nums[j])
    # positive or negative
    if x > 0 :
        pos+=1 
    elif x < 0 :
        neg+=1 
    
    # odd or even
    if x % 2 !=0 :
        odd+=1 
    else:
        even+=1 
    

print("Even:",even)
print("Odd:",odd)
print("Positive:",pos)
print("Negative:",neg)
```
