Conditionals statements in Python

```python
raining = True
muddy = True

if raining == True:
    print("Rain coat is needed..!")
    if muddy == True:
        print("Sandal is needed..!")
    else:
        print("Bring sneakers..!")
else:
    print("Go without a rain coat..!")
```
