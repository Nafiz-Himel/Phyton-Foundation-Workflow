```python
#List Comprehension with List of String
fruits = ["apple", "mango", "banana"]

capitalize_fruits = []
for fruit in fruits:
    capitalize_fruits.append(fruit.upper())

print(capitalize_fruits)

capitalize_fruits = [fruit.lower() for fruit in fruits]
print(capitalize_fruits)
```

E. Lowest Number

```python
n = int(input())
num = input().split()
# print(nums,type(nums))
min = 100005
idx = 0
for i in range(n):
    if min > int(num[i]):
        min = int(num[i])
        idx = i
print(f"{min} {idx+1}")
```
