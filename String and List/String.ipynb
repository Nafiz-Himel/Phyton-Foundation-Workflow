String Methods

```python
string = "Nafiz himel Rayhan Himel"
# processed_string = string.lower()
processed_string = string.upper()

# Length
length = len(string)
print(length,type(length))
print(processed_string)
print("himel" in processed_string) #bool

# shurur kon index theke substring ta shuru hoise 
shuru_index = processed_string.find("HIMEL")
print(shuru_index)

# shesher dike kon index theke substring ta shuru hoise
sesh_index = processed_string.rfind("HIMEL")
print(sesh_index)

#Substring koybar ache
count = processed_string.count("HIMEL")
print(count)

#replace substring
processed_string = string.replace("Himel","himel")
print(processed_string)

print(string) #immutable structure
```
