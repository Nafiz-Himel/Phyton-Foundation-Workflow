Tuple

```python
# declaration 

tup = (10,20,30,40) 

float_tup = (10.5,10.2)
mixed_tup = (10 , 10.5 , "str", True) 


lst = [10,20,30]

tup1 = tuple(lst)

print(type(tup1), tup1)

tup_int = (100)
# tup_with_one_val = tuple([100])
tup_with_one_val = (100,)
print(type(tup_int),tup_int)
print(type(tup_with_one_val),tup_with_one_val)
```

```python
#access
tup = (10 , 20 , 30 , 40 , 10.50, "str" , False)

# index
print(tup[4])

new_tup = tup[0:3]
print(new_tup)
```

```python
#immutablity
lst = [10,20,30,40]

lst.append(60)
lst[1] = 100
print(lst)

#immutability
tup = (10,20,30,40)

tup[1] = 100

print(tup)
```
