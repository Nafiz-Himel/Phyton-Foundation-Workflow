Set Mathematical Methods

```python
a = {1,2,3}
b = {1,2,3,2,4,5}

print(a.union(b))
print(a.intersection(b))

print(a.isdisjoint(b))

print(a.issubset(b))
```

Dictionaries in Python

```python
dic = {}
print(type(dic))

dic = {
    "name" : "himel",
    "age" : 20,
    "height" : 5.5,
    "address" : "Dhaka",
    "numbers" : [10,20,30]
}
print(type(dic),dic)
```

```python
# access

#dic_name[key]
print(dic["name"],dic["numbers"])

print(dic.get("age"))

dic["name"] = "nafiz"
print(dic)


dic = {
    "name" : "himel",
    "age" : 20, #50 diye replace hoye gese
    "height" : 5.5,
    "address" : "Dhaka",
    "numbers" : [10,20,30],
    "age" : 50
}
print(dic)
```

Dictionary Methods

```python
#printing default values
print(dic)

print(dic.get("math_amrks"))
print(dic.get("math_amrks",0))
```

```python
dic["math_marks"] = 100
print(dic)
dic.update({"eng_amrks": 40})

del dic["math_marks"]
print(dic)

dic_2 = dic.copy()
print(dic_2)
```

```python
# testing immutability
dic_3 = {
    [10,20,30] : "a"
}
print(dic_3)

# dic_3 = {
#     (10,20,30) : "a"
# }
# print(dic_3)
```

Dictionary View Methods

```python
keys = dic.keys()
print(keys,type(keys))

dic["math_marks"] = 100
print(keys)


values = dic.values()
print(values)

dic["math_marks"] = 75
print(values)


items = dic.items()
print(items,type(items))

dic["math_marks"] = 33
print(items)
```

```python
# iteration

for key,value in dic:
    print(key,value)
```

```python
for key,value in dic.items():
    print(key,type(key),value,type(value))
```
